## 2. Closure pull definition

The current closure pull is defined as

\[
\mathrm{pull}=
\frac{A_{\mathrm{obs}}-A_{\mathrm{pred}}}
{\sqrt{\sigma_{A,\mathrm{obs}}^2+\sigma_{A,\mathrm{pred}}^2}},
\qquad
A_{\mathrm{pred}}=\frac{BC}{D}
\]

The A_obs uncertainty comes from the histogram variance, and the prediction uncertainty is propagated assuming independent B/C/D regions.

\[
\left(\frac{\sigma_{A,\mathrm{pred}}}{A_{\mathrm{pred}}}\right)^2
=
\left(\frac{\sigma_B}{B}\right)^2+
\left(\frac{\sigma_C}{C}\right)^2+
\left(\frac{\sigma_D}{D}\right)^2
\]

The current pull excludes systematic uncertainties and covariance between regions. It is a statistical diagnostic, not a final significance.

## 9. Pilot conclusions

### Supported conclusions

1. Blinding, SR/VR histogram access, and ABCD calculations work.
2. Current VR Data have large statistical pulls in both topologies, indicating insufficient closure.
3. SR MC kappa is statistically unstable and does not provide a reliable basis for a correction.
4. The raw SR Data B×C/D candidate can be calculated, but its systematic prescription is not finalized.
5. Leakage into the SR ABCD sidebands is a more direct signal-contamination diagnostic than migration into the tight VR.
6. The 2mu2e C-region leakage and signal ABCD bias deserve priority in follow-up studies.

### Unsupported conclusions

1. The observed SR Data A yield and discovery significance
2. A final kappa correction or final background estimate
3. Signal selection efficiency based on a generated-event denominator
4. Total SR-to-VR migration, including the tight VR
5. Acceptance loss attributed to individual causes in mixed displacement/mass/DeltaPhi failures
6. A claim of physically zero leakage at zero-yield points
