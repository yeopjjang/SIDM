## 2. Closure pull definition

현재 closure pull은

\[
\mathrm{pull}=
\frac{A_{\mathrm{obs}}-A_{\mathrm{pred}}}
{\sqrt{\sigma_{A,\mathrm{obs}}^2+\sigma_{A,\mathrm{pred}}^2}},
\qquad
A_{\mathrm{pred}}=\frac{BC}{D}
\]

로 계산한다. A_obs uncertainty는 histogram variance에서 얻고, prediction uncertainty는 B/C/D가 독립이라는 가정 아래 전파한다.

\[
\left(\frac{\sigma_{A,\mathrm{pred}}}{A_{\mathrm{pred}}}\right)^2
=
\left(\frac{\sigma_B}{B}\right)^2+
\left(\frac{\sigma_C}{C}\right)^2+
\left(\frac{\sigma_D}{D}\right)^2
\]

현재 pull에는 systematic uncertainty와 region 간 covariance가 포함되지 않는다. 따라서 순수 statistical diagnostic이며 final significance가 아니다.

## 9. Pilot conclusions

### 현재 말할 수 있는 것

1. Blinding, SR/VR histogram 접근과 ABCD 계산 machinery가 동작한다.
2. 현재 VR Data는 두 topology 모두 큰 statistical pull을 보여 closure가 충분하지 않다.
3. SR MC kappa는 통계적으로 불안정해 correction 근거로 쓰기 어렵다.
4. SR Data raw B×C/D 후보는 계산 가능하지만 final systematic이 없다.
5. Tight-VR 유입보다 SR ABCD sideband leakage가 더 직접적인 signal contamination diagnostic이다.
6. 2mu2e C-region leakage와 signal ABCD bias를 우선 추적할 가치가 있다.

### 현재 말할 수 없는 것

1. SR Data A observed yield와 discovery significance
2. Final kappa correction 또는 final background estimate
3. Generated denominator 기반 signal selection efficiency
4. Tight VR을 포함한 전체 SR-to-VR migration
5. Mixed displacement/mass/DeltaPhi failure의 원인별 acceptance loss
6. Zero-yield point의 물리적 zero leakage 주장
