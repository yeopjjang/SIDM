```python
import json
import os
from collections import defaultdict

input_dir  = "/home/cms-jovyan/workspace/CMS-SIDM-PR/sidm/studies/JSON"
output_dir = "/home/cms-jovyan/workspace/CMS-SIDM-PR/sidm/studies/JSON/Merged"

processed_files = [
    "processed_lumis_DoubleMuon_2018A_0.json",
    "processed_lumis_DoubleMuon_2018A_1.json",
    "processed_lumis_DoubleMuon_2018A_2.json",

    "processed_lumis_DoubleMuon_2018B_0.json",
    "processed_lumis_DoubleMuon_2018B_1.json",
    "processed_lumis_DoubleMuon_2018B_2.json",
    "processed_lumis_DoubleMuon_2018B_3.json",
    "processed_lumis_DoubleMuon_2018B_4.json",
    "processed_lumis_DoubleMuon_2018B_5.json",
    "processed_lumis_DoubleMuon_2018B_6.json",
    "processed_lumis_DoubleMuon_2018B_7.json",
    "processed_lumis_DoubleMuon_2018B_8.json",
    "processed_lumis_DoubleMuon_2018B_9.json",
    "processed_lumis_DoubleMuon_2018B_10.json",
    "processed_lumis_DoubleMuon_2018B_11.json",
    "processed_lumis_DoubleMuon_2018B_12.json",
    "processed_lumis_DoubleMuon_2018B_13.json",
    
    "processed_lumis_DoubleMuon_2018C.json",

    "processed_lumis_DoubleMuon_2018D_0.json",
    "processed_lumis_DoubleMuon_2018D_1.json",
    "processed_lumis_DoubleMuon_2018D_2.json",
    "processed_lumis_DoubleMuon_2018D_3.json",
    "processed_lumis_DoubleMuon_2018D_4.json",
    "processed_lumis_DoubleMuon_2018D_5.json",
    "processed_lumis_DoubleMuon_2018D_6.json",
    "processed_lumis_DoubleMuon_2018D_7.json",
    "processed_lumis_DoubleMuon_2018D_8.json",
    "processed_lumis_DoubleMuon_2018D_9.json",
    "processed_lumis_DoubleMuon_2018D_10.json",
    "processed_lumis_DoubleMuon_2018D_11.json",
    "processed_lumis_DoubleMuon_2018D_12.json",
    "processed_lumis_DoubleMuon_2018D_13.json",
    "processed_lumis_DoubleMuon_2018D_14.json",
    "processed_lumis_DoubleMuon_2018D_15.json",
    "processed_lumis_DoubleMuon_2018D_16.json",
    "processed_lumis_DoubleMuon_2018D_17.json",
    "processed_lumis_DoubleMuon_2018D_18.json",
    "processed_lumis_DoubleMuon_2018D_19.json",
    "processed_lumis_DoubleMuon_2018D_20.json",
    "processed_lumis_DoubleMuon_2018D_21.json",
    "processed_lumis_DoubleMuon_2018D_22.json",
    "processed_lumis_DoubleMuon_2018D_23.json",
    "processed_lumis_DoubleMuon_2018D_24.json",
    "processed_lumis_DoubleMuon_2018D_25.json",
    "processed_lumis_DoubleMuon_2018D_26.json",
    "processed_lumis_DoubleMuon_2018D_27.json",
    "processed_lumis_DoubleMuon_2018D_28.json",
    "processed_lumis_DoubleMuon_2018D_29.json",
    "processed_lumis_DoubleMuon_2018D_30.json",
    "processed_lumis_DoubleMuon_2018D_31.json",
    "processed_lumis_DoubleMuon_2018D_32.json",
    "processed_lumis_DoubleMuon_2018D_33.json",
    "processed_lumis_DoubleMuon_2018D_34.json",
    "processed_lumis_DoubleMuon_2018D_35.json",
    "processed_lumis_DoubleMuon_2018D_36.json",
    "processed_lumis_DoubleMuon_2018D_37.json",
    "processed_lumis_DoubleMuon_2018D_38.json",
    "processed_lumis_DoubleMuon_2018D_39.json",
    "processed_lumis_DoubleMuon_2018D_40.json",
    "processed_lumis_DoubleMuon_2018D_41.json",
    "processed_lumis_DoubleMuon_2018D_42.json",
    "processed_lumis_DoubleMuon_2018D_43.json",
    "processed_lumis_DoubleMuon_2018D_44.json",
    "processed_lumis_DoubleMuon_2018D_45.json",
    "processed_lumis_DoubleMuon_2018D_46.json",
    "processed_lumis_DoubleMuon_2018D_47.json",
    "processed_lumis_DoubleMuon_2018D_48.json",
    "processed_lumis_DoubleMuon_2018D_49.json",
]

def expand_ranges(ranges):
    lumis = set()
    for start, end in ranges:
        lumis.update(range(start, end + 1))
    return lumis

def compress_lumis(lumis):
    lumis = sorted(lumis)
    if not lumis:
        return []

    merged = []
    start = prev = lumis[0]

    for x in lumis[1:]:
        if x == prev + 1:
            prev = x
        else:
            merged.append([start, prev])
            start = prev = x
    merged.append([start, prev])
    return merged

# ----------------------------------
# 3. merge
# ----------------------------------
all_lumis = defaultdict(set)

for fname in processed_files:
    full_path = os.path.join(input_dir, fname)

    with open(full_path) as f:
        data = json.load(f)

    for run, ranges in data.items():
        all_lumis[str(run)].update(expand_ranges(ranges))

# ----------------------------------
# 4. output
# ----------------------------------
merged_json = {
    run: compress_lumis(lumis)
    for run, lumis in sorted(all_lumis.items(), key=lambda x: int(x[0]))
}

output_name = "processed_lumis_DoubleMuon_2018ABCD_merged.json"
output_path = os.path.join(output_dir, output_name)

with open(output_path, "w") as f:
    json.dump(merged_json, f, indent=2)

print(f"Saved: {output_path}")
print(f"Number of runs: {len(merged_json)}")
```
