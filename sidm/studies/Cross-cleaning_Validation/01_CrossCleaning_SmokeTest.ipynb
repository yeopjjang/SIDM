```python
# ============================================================
# 3. Histogram access helpers
# ============================================================
import numpy as np
import matplotlib.pyplot as plt
import os


def get_hist_store(out, sample):
    if sample not in out:
        raise KeyError(
            f"Sample {sample!r} was not found.\n"
            f"Available keys: {list(out.keys())[:30]}"
        )

    sample_output = out[sample]

    if not isinstance(sample_output, dict):
        raise TypeError(
            f"out[{sample!r}] is not a dictionary. "
            f"Found type: {type(sample_output)}"
        )

    if "hists" in sample_output:
        return sample_output["hists"]

    return sample_output


def get_histogram(out, sample, histname):
    hist_store = get_hist_store(out, sample)

    if histname not in hist_store:
        raise KeyError(
            f"Histogram {histname!r} was not found for sample {sample!r}."
        )

    return hist_store[histname]


def get_axis_names(histogram):
    return [axis.name for axis in histogram.axes]


def get_axis_categories(histogram, axis_name):
    axis = histogram.axes[axis_name]
    return list(axis)


def slice_histogram_channel(histogram, channel):
    axis_names = get_axis_names(histogram)

    if "channel" in axis_names:
        return histogram[{"channel": channel}]

    for axis in histogram.axes:
        try:
            categories = list(axis)
        except Exception:
            continue

        if channel in categories:
            return histogram[{axis.name: channel}]

    raise KeyError(
        f"Could not find channel {channel!r} in histogram.\n"
        f"Histogram axes: {axis_names}"
    )


def reduce_to_1d(histogram):
    h = histogram

    while len(h.axes) > 1:
        categorical_axis_found = False

        for axis in h.axes:
            try:
                categories = list(axis)
            except Exception:
                categories = []

            if len(categories) > 0:
                h = h[{axis.name: sum}]
                categorical_axis_found = True
                break

        if not categorical_axis_found:
            raise ValueError(
                "Histogram still contains more than one numeric axis.\n"
                f"Axes: {[axis.name for axis in h.axes]}"
            )

    return h


def histogram_errors(histogram):
    values = np.asarray(histogram.values(), dtype=float)

    try:
        variances = histogram.variances()
    except Exception:
        variances = None

    if variances is None:
        return np.sqrt(np.abs(values))

    variances = np.asarray(variances, dtype=float)
    variances = np.where(np.isfinite(variances) & (variances >= 0), variances, 0.0)

    return np.sqrt(variances)


def extract_1d_values_and_errors(out, sample, histname, channel):
    histogram = get_histogram(
        out=out,
        sample=sample,
        histname=histname,
    )

    histogram = slice_histogram_channel(
        histogram=histogram,
        channel=channel,
    )

    histogram = reduce_to_1d(histogram)

    values = np.asarray(histogram.values(), dtype=float)
    errors = histogram_errors(histogram)
    edges = np.asarray(histogram.axes[0].edges, dtype=float)

    return values, errors, edges


def extract_1d_values(out, sample, histname, channel):
    values, _, edges = extract_1d_values_and_errors(
        out=out,
        sample=sample,
        histname=histname,
        channel=channel,
    )

    return values, edges


def bin_centers(edges):
    return 0.5 * (edges[:-1] + edges[1:])


def normalize_values_and_errors(values, errors):
    integral = np.sum(values)

    if integral <= 0:
        return np.zeros_like(values), np.zeros_like(errors)

    return values / integral, errors / integral



def weighted_efficiency_errors(numerator, denominator, denominator_error):
    efficiency = safe_ratio(numerator, denominator)
    den_var = denominator_error ** 2
    error = np.zeros_like(efficiency, dtype=float)

    valid = (
        np.isfinite(efficiency)
        & (denominator > 0)
        & (den_var > 0)
    )

    n_eff = np.zeros_like(efficiency, dtype=float)
    n_eff[valid] = denominator[valid] ** 2 / den_var[valid]

    clipped_eff = np.clip(efficiency, 0.0, 1.0)
    valid = valid & (n_eff > 0)
    error[valid] = np.sqrt(clipped_eff[valid] * (1.0 - clipped_eff[valid]) / n_eff[valid])

    error[~np.isfinite(error)] = 0.0
    return error
```

```python
# ============================================================
# 4. PF × DSA category definitions
# ============================================================
pf_categories = [
    ("nosel", "PF: no selection"),
    ("noPF", "PF = 0"),
    ("onePF", "PF = 1"),
    ("twoPF", "PF = 2"),
]

dsa_categories = [
    ("nosel", "DSA: no selection"),
    ("noDSA", "DSA = 0"),
    ("oneDSA", "DSA = 1"),
    ("twoDSA", "DSA = 2"),
]


def make_pf_dsa_histname(
    variable,
    pf_category,
    dsa_category,
    trailing_suffix="",
):

    if pf_category == "nosel" and dsa_category == "nosel":
        category_suffix = "nosel"

    elif pf_category == "nosel":
        category_suffix = dsa_category

    elif dsa_category == "nosel":
        category_suffix = pf_category

    else:
        category_suffix = f"{pf_category}_{dsa_category}"

    return f"{variable}_{category_suffix}{trailing_suffix}"


def build_pf_dsa_histname_grid(variable, trailing_suffix=""):
    """
    Build a 4×4 list of histogram names.
    """

    histname_grid = []

    for pf_category, _ in pf_categories:
        row = []

        for dsa_category, _ in dsa_categories:
            histname = make_pf_dsa_histname(
                variable=variable,
                pf_category=pf_category,
                dsa_category=dsa_category,
                trailing_suffix=trailing_suffix,
            )

            row.append(histname)

        histname_grid.append(row)

    return histname_grid


test_grid = build_pf_dsa_histname_grid(
    variable="genAs_toMu_lxy",
)

for row in test_grid:
    print(row)
```

```python
# ============================================================
# 5. Check required histograms
# ============================================================
def check_pf_dsa_histograms(
    out,
    sample,
    variable,
    trailing_suffix="",
):
    hist_store = get_hist_store(out, sample)

    histname_grid = build_pf_dsa_histname_grid(
        variable=variable,
        trailing_suffix=trailing_suffix,
    )

    available = []
    missing = []

    for row in histname_grid:
        for histname in row:
            if histname in hist_store:
                available.append(histname)
            else:
                missing.append(histname)

    print("Variable:", f"{variable}{trailing_suffix}")
    print("Available:", len(available))
    print("Missing:", len(missing))

    if missing:
        print("\nMissing histograms:")
        for histname in missing:
            print("  -", histname)

    return missing


missing_2mu2e = check_pf_dsa_histograms(
    out=out_2mu2e,
    sample=sample_2mu2e,
    variable="genAs_toMu_lxy",
)

missing_4mu = check_pf_dsa_histograms(
    out=out_4mu,
    sample=sample_4mu,
    variable="genAs_toMu_lxy",
)
```

```python
# ============================================================
# 6. Numerical helper
# ============================================================
def safe_ratio(numerator, denominator):
    numerator = np.asarray(numerator, dtype=float)
    denominator = np.asarray(denominator, dtype=float)

    if numerator.shape != denominator.shape:
        raise ValueError(
            f"Numerator and denominator shapes differ: "
            f"{numerator.shape} vs {denominator.shape}"
        )

    ratio = np.full(
        numerator.shape,
        np.nan,
        dtype=float,
    )

    valid = (
        np.isfinite(numerator)
        & np.isfinite(denominator)
        & (denominator > 0)
    )

    ratio[valid] = numerator[valid] / denominator[valid]

    return ratio
```

```python
# ============================================================
# 7. Main PF × DSA 4×4 plotting function
# ============================================================
def plot_pf_dsa_4x4_overlay(
    out,
    sample,
    variable,
    trailing_suffix="",
    channels=channels,
    mode="ratio_to_nosel",
    xlabel=None,
    sample_label=None,
    output_dir="plots/pf_dsa_4x4",
    output_name=None,
    figsize=(40, 40),
    xlim=None,
    ylim=None,
    logy=False,
    show_legend=True,
    show_errors=True,
):

    valid_modes = [
        "ratio_to_nosel",
        "shape",
        "yield",
    ]

    if mode not in valid_modes:
        raise ValueError(
            f"Invalid mode '{mode}'. "
            f"Choose one of: {valid_modes}"
        )

    hist_store = get_hist_store(out, sample)

    histname_grid = build_pf_dsa_histname_grid(
        variable=variable,
        trailing_suffix=trailing_suffix,
    )

    # --------------------------------------------------------
    # Check histogram availability
    # --------------------------------------------------------

    missing_histograms = []

    for row in histname_grid:
        for histname in row:
            if histname not in hist_store:
                missing_histograms.append(histname)

    if missing_histograms:
        print("Cannot make plot because histograms are missing:")

        for histname in missing_histograms:
            print("  -", histname)

        return None

    # --------------------------------------------------------
    # Read no-selection denominator for every channel
    # --------------------------------------------------------

    denominator_values = {}
    denominator_errors = {}
    denominator_edges = {}

    if mode == "ratio_to_nosel":
        nosel_histname = histname_grid[0][0]

        for channel in channels:
            values, errors, edges = extract_1d_values_and_errors(
                out=out,
                sample=sample,
                histname=nosel_histname,
                channel=channel,
            )

            denominator_values[channel] = values
            denominator_errors[channel] = errors
            denominator_edges[channel] = edges

    # --------------------------------------------------------
    # Create canvas
    # --------------------------------------------------------

    fig, axes = plt.subplots(
        nrows=4,
        ncols=4,
        figsize=figsize,
        sharex=True,
        sharey=True,
    )

    # --------------------------------------------------------
    # Draw panels
    # --------------------------------------------------------

    for pf_index, (_, pf_label) in enumerate(pf_categories):
        for dsa_index, (_, dsa_label) in enumerate(dsa_categories):

            ax = axes[pf_index, dsa_index]
            histname = histname_grid[pf_index][dsa_index]

            for channel in channels:
                values, errors, edges = extract_1d_values_and_errors(
                    out=out,
                    sample=sample,
                    histname=histname,
                    channel=channel,
                )

                if mode == "ratio_to_nosel":
                    denominator = denominator_values[channel]

                    if not np.allclose(
                        edges,
                        denominator_edges[channel],
                    ):
                        raise ValueError(
                            f"Bin edges differ between '{histname}' "
                            f"and the no-selection histogram for "
                            f"channel '{channel}'."
                        )

                    plot_values = safe_ratio(
                        numerator=values,
                        denominator=denominator,
                    )
                    plot_errors = weighted_efficiency_errors(
                        numerator=values,
                        denominator=denominator,
                        denominator_error=denominator_errors[channel],
                    )

                elif mode == "shape":
                    integral = np.sum(values)

                    plot_values, plot_errors = normalize_values_and_errors(
                        values,
                        errors,
                    )

                else:
                    plot_values = values
                    plot_errors = errors

                color = channel_colors.get(channel, None)

                ax.stairs(
                    values=plot_values,
                    edges=edges,
                    label=channel_labels.get(channel, channel),
                    color=color,
                    linestyle=channel_linestyles.get(channel, "-"),
                    linewidth=2.0,
                )

                if show_errors:
                    ax.errorbar(
                        bin_centers(edges),
                        plot_values,
                        yerr=plot_errors,
                        fmt="none",
                        ecolor=color,
                        elinewidth=1.0,
                        capsize=0,
                        alpha=0.7,
                    )

            # Horizontal reference line for ratio plots
            if mode == "ratio_to_nosel":
                ax.axhline(
                    y=1.0,
                    color="gray",
                    linestyle="--",
                    linewidth=1.0,
                    alpha=0.7,
                )

            ax.set_title(
                f"{pf_label}\n{dsa_label}",
                fontsize=30,
            )

            ax.grid(
                True,
                which="both",
                linestyle=":",
                alpha=0.3,
            )

            if xlim is not None:
                ax.set_xlim(xlim)

            if ylim is not None:
                ax.set_ylim(ylim)

            if logy and mode != "ratio_to_nosel":
                ax.set_yscale("log")

            if pf_index == 3:
                ax.set_xlabel(
                    xlabel if xlabel is not None else variable,
                    fontsize=25,
                )

            if dsa_index == 0:
                if mode == "ratio_to_nosel":
                    ylabel = "Category / No selection"
                elif mode == "shape":
                    ylabel = "Normalized events"
                else:
                    ylabel = "Yield"

                ax.set_ylabel(
                    ylabel,
                    fontsize=25,
                )

    # --------------------------------------------------------
    # Common legend
    # --------------------------------------------------------

    if show_legend:
        handles, labels = axes[0, 0].get_legend_handles_labels()

        fig.legend(
            handles=handles,
            labels=labels,
            loc="upper center",
            bbox_to_anchor=(0.5, 0.955),
            ncol=len(channels),
            frameon=False,
            fontsize=25,
        )

    # --------------------------------------------------------
    # Figure title
    # --------------------------------------------------------

    if sample_label is None:
        sample_label = sample

    mode_titles = {
        "ratio_to_nosel": "Category / No selection",
        "shape": "Unit-area normalized",
        "yield": "Yield",
    }

    displayed_variable = f"{variable}{trailing_suffix}"

    fig.suptitle(
        f"{sample_label}\n"
        f"{displayed_variable}: {mode_titles[mode]}",
        fontsize=25,
        y=0.995,
    )

    plt.tight_layout(
        rect=[0.02, 0.02, 0.98, 0.91],
    )

    # --------------------------------------------------------
    # Save
    # --------------------------------------------------------

    os.makedirs(
        output_dir,
        exist_ok=True,
    )

    if output_name is None:
        safe_sample = sample.replace("/", "_")

        output_name = (
            f"{safe_sample}_"
            f"{displayed_variable}_"
            f"pf_dsa_4x4_"
            f"{mode}.png"
        )

    output_path = os.path.join(
        output_dir,
        output_name,
    )

    fig.savefig(
        output_path,
        dpi=200,
        bbox_inches="tight",
    )

    plt.show()
    plt.close(fig)

    print("Saved:", output_path)

    return output_path
```

```python
# ============================================================
# Inclusive 1D channel overlay without ratio panel
# ============================================================

def plot_inclusive_channel_overlay_no_ratio(
    out,
    sample,
    histname,
    channels=channels,
    xlabel=None,
    ylabel="Yield",
    sample_label=None,
    output_dir="plots/inclusive",
    output_name=None,
    normalize=False,
    logy=False,
    xlim=None,
    ylim=None,
    vertical_lines=None,
    figsize=(16, 12),
    show_errors=True,
):
    values_by_channel = {}
    errors_by_channel = {}
    edges_by_channel = {}

    for channel in channels:
        values, errors, edges = extract_1d_values_and_errors(
            out=out,
            sample=sample,
            histname=histname,
            channel=channel,
        )

        values = np.asarray(values, dtype=float)
        errors = np.asarray(errors, dtype=float)
        edges = np.asarray(edges, dtype=float)

        if normalize:
            values, errors = normalize_values_and_errors(values, errors)

        values_by_channel[channel] = values
        errors_by_channel[channel] = errors
        edges_by_channel[channel] = edges

    reference_edges = edges_by_channel[channels[0]]

    for channel in channels:
        if not np.allclose(edges_by_channel[channel], reference_edges):
            raise ValueError(f"Bin edges differ for channel {channel!r}.")

    fig, ax = plt.subplots(figsize=figsize)
    ax.text(0.0, 1.01, "Private work (CMS simulation)", transform=ax.transAxes, ha="left", va="bottom")
    centers = bin_centers(reference_edges)

    for channel in channels:
        color = channel_colors.get(channel)

        ax.stairs(
            values_by_channel[channel],
            reference_edges,
            label=channel_labels.get(channel, channel),
            color=color,
            linestyle=channel_linestyles.get(channel, "-"),
            linewidth=2.0,
        )

        if show_errors:
            ax.errorbar(
                centers,
                values_by_channel[channel],
                yerr=errors_by_channel[channel],
                fmt="o",
                markersize=4,
                color=color,
                ecolor=color,
                elinewidth=1.2,
                capsize=2,
                alpha=0.85,
            )

    if vertical_lines is not None:
        for x_position in vertical_lines:
            ax.axvline(
                x_position,
                color="gray",
                linestyle="--",
                linewidth=1.2,
                alpha=0.8,
            )

    if logy:
        ax.set_yscale("log")

    if xlim is not None:
        ax.set_xlim(xlim)

    if ylim is not None:
        ax.set_ylim(ylim)

    ax.set_xlabel(xlabel if xlabel is not None else histname)
    ax.set_ylabel("Normalized events" if normalize else ylabel)
    ax.grid(True, which="both", linestyle=":", alpha=0.3)
    ax.legend(frameon=False, fontsize=25)

    if sample_label is None:
        sample_label = sample

    plot_type = "Normalized shape" if normalize else "Yield"
    ax.set_title(f"{sample_label}\n{plot_type}", fontsize=25)

    os.makedirs(output_dir, exist_ok=True)

    if output_name is None:
        suffix = "shape" if normalize else "yield"
        output_name = f"{sample}_{histname}_channel_overlay_{suffix}_no_ratio.png"

    output_path = os.path.join(output_dir, output_name)
    fig.savefig(output_path, dpi=200, bbox_inches="tight")
    plt.show()
    plt.close(fig)

    print("Saved:", output_path)
    return output_path
```

```python
# ============================================================
# 18. Gen-muon matching efficiency versus dark-photon Lxy
# ============================================================

def plot_dsa_id_efficiency(
    out,
    sample,
    channels=channels,
    denominator_histname="dsa_id_eff_genMu_fromA_lxy_den",
    numerator_histname="dsa_id_eff_genMu_fromA_lxy_selectedDsa_num",
    sample_label=None,
    output_dir="plots/dsa_efficiency",
    output_name=None,
    output_tag="dsa_id_efficiency",
    y_label="Selected DSA matching efficiency",
    xlim=(0, 400),
    ylim=(0, 1.1),
    figsize=(16, 12),
):
    """
    Plot weighted reco-muon matching efficiency versus gen dark-photon Lxy.

    Efficiency = reco-matched gen-muon yield / gen-muon denominator.
    """

    fig, ax = plt.subplots(
        figsize=figsize,
    )
    ax.text(0.0, 1.01, "Private work (CMS simulation)", transform=ax.transAxes, ha="left", va="bottom")

    reference_edges = None

    for channel in channels:
        denominator, denominator_error, denominator_edges = extract_1d_values_and_errors(
            out=out,
            sample=sample,
            histname=denominator_histname,
            channel=channel,
        )

        numerator, _, numerator_edges = extract_1d_values_and_errors(
            out=out,
            sample=sample,
            histname=numerator_histname,
            channel=channel,
        )

        if not np.allclose(
            denominator_edges,
            numerator_edges,
        ):
            raise ValueError(
                f"Numerator and denominator binning differ "
                f"for channel {channel!r}."
            )

        efficiency = safe_ratio(
            numerator,
            denominator,
        )

        efficiency_error = weighted_efficiency_errors(
            numerator=numerator,
            denominator=denominator,
            denominator_error=denominator_error,
        )

        ax.errorbar(
            bin_centers(denominator_edges),
            efficiency,
            yerr=efficiency_error,
            label=channel_labels.get(channel, channel),
            color=channel_colors.get(channel),
            marker="o",
            linestyle="none",
            markersize=5,
            elinewidth=1.2,
            capsize=2,
        )

        reference_edges = denominator_edges

    ax.set_xlabel(
        r"Gen $Z_{d}\rightarrow\mu\mu$ $L_{xy}$ [cm]",
    )

    ax.set_ylabel(
        y_label,
    )

    ax.set_xlim(xlim)
    ax.set_ylim(ylim)

    ax.axhline(
        1.0,
        color="gray",
        linestyle="--",
        linewidth=1.0,
    )

    ax.grid(
        True,
        which="both",
        linestyle=":",
        alpha=0.3,
    )

    ax.legend(
        frameon=False,
    )

    if sample_label is None:
        sample_label = sample

    ax.set_title(
        sample_label,
    )

    os.makedirs(
        output_dir,
        exist_ok=True,
    )

    if output_name is None:
        output_name = (
            f"{sample}_{output_tag}_vs_lxy.png"
        )

    output_path = os.path.join(
        output_dir,
        output_name,
    )

    fig.savefig(
        output_path,
        dpi=200,
        bbox_inches="tight",
    )

    plt.show()
    plt.close(fig)

    print("Saved:", output_path)

    return output_path



def plot_muon_efficiency_overlay(
    out,
    sample,
    channel="base_finalCC",
    denominator_histname="dsa_id_eff_genMu_fromA_lxy_den",
    numerator_histnames=None,
    sample_label=None,
    output_dir="plots/muon_efficiency_overlay",
    output_name=None,
    xlim=(0, 400),
    ylim=(0, 1.1),
    figsize=(16, 12),
):
    if numerator_histnames is None:
        numerator_histnames = {
            "PF": "dsa_id_eff_genMu_fromA_lxy_selectedPf_num",
            "DSA": "dsa_id_eff_genMu_fromA_lxy_selectedDsa_num",
            "PF or DSA": "dsa_id_eff_genMu_fromA_lxy_selectedPfOrDsa_num",
        }

    denominator, denominator_error, denominator_edges = extract_1d_values_and_errors(
        out=out,
        sample=sample,
        histname=denominator_histname,
        channel=channel,
    )

    colors = {
        "PF": "tab:blue",
        "DSA": "tab:orange",
        "PF or DSA": "tab:green",
    }

    fig, ax = plt.subplots(figsize=figsize)
    ax.text(0.0, 1.01, "Private work (CMS simulation)", transform=ax.transAxes, ha="left", va="bottom")

    for label, numerator_histname in numerator_histnames.items():
        numerator, _, numerator_edges = extract_1d_values_and_errors(
            out=out,
            sample=sample,
            histname=numerator_histname,
            channel=channel,
        )

        if not np.allclose(denominator_edges, numerator_edges):
            raise ValueError(f"Numerator and denominator binning differ for {label!r}.")

        efficiency = safe_ratio(numerator, denominator)
        efficiency_error = weighted_efficiency_errors(
            numerator=numerator,
            denominator=denominator,
            denominator_error=denominator_error,
        )

        ax.errorbar(
            bin_centers(denominator_edges),
            efficiency,
            yerr=efficiency_error,
            label=label,
            color=colors.get(label),
            marker="o",
            linestyle="none",
            markersize=5,
            elinewidth=1.2,
            capsize=2,
        )

    ax.axhline(1.0, color="gray", linestyle="--", linewidth=1.0)
    ax.set_xlabel(r"Gen $Z_{d}\rightarrow\mu\mu$ $L_{xy}$ [cm]")
    ax.set_ylabel("Matching efficiency")
    ax.set_xlim(xlim)
    ax.set_ylim(ylim)
    ax.grid(True, which="both", linestyle=":", alpha=0.3)
    ax.legend(frameon=False)

    if sample_label is None:
        sample_label = sample

    ax.set_title(f"{sample_label}\n{channel_labels.get(channel, channel)}")
    os.makedirs(output_dir, exist_ok=True)

    if output_name is None:
        output_name = f"{sample}_{channel}_pf_dsa_general_efficiency_overlay.png"

    output_path = os.path.join(output_dir, output_name)
    fig.savefig(output_path, dpi=200, bbox_inches="tight")
    plt.show()
    plt.close(fig)

    print("Saved:", output_path)
    return output_path
```

```python
# ============================================================
# 19. Extract a channel-sliced 2D histogram
# ============================================================

def reduce_to_2d(histogram):
    """
    Sum over remaining categorical axes until exactly two axes remain.
    """

    h = histogram

    while len(h.axes) > 2:
        categorical_axis_found = False

        for axis in h.axes:
            try:
                categories = list(axis)
            except Exception:
                categories = []

            if len(categories) > 0:
                h = h[{axis.name: sum}]
                categorical_axis_found = True
                break

        if not categorical_axis_found:
            raise ValueError(
                "Histogram contains more than two numeric axes.\n"
                f"Axes: {[axis.name for axis in h.axes]}"
            )

    if len(h.axes) != 2:
        raise ValueError(
            f"Expected a 2D histogram, but found {len(h.axes)} axes."
        )

    return h


def extract_2d_values_and_errors(
    out,
    sample,
    histname,
    channel,
):
    histogram = get_histogram(
        out=out,
        sample=sample,
        histname=histname,
    )

    histogram = slice_histogram_channel(
        histogram=histogram,
        channel=channel,
    )

    histogram = reduce_to_2d(
        histogram,
    )

    values = np.asarray(
        histogram.values(),
        dtype=float,
    )

    errors = histogram_errors(histogram)

    x_edges = np.asarray(
        histogram.axes[0].edges,
        dtype=float,
    )

    y_edges = np.asarray(
        histogram.axes[1].edges,
        dtype=float,
    )

    return values, errors, x_edges, y_edges


def extract_2d_values(
    out,
    sample,
    histname,
    channel,
):
    values, _, x_edges, y_edges = extract_2d_values_and_errors(
        out=out,
        sample=sample,
        histname=histname,
        channel=channel,
    )
    return values, x_edges, y_edges


def extract_segment_fraction_projection_for_dr_cut(
    out,
    sample,
    channel,
    histname="pf_dsa_cc_cut_plane_numMatchGe1",
    dr_max=0.2,
):
    values, errors, dr_edges, segment_fraction_edges = extract_2d_values_and_errors(
        out=out,
        sample=sample,
        histname=histname,
        channel=channel,
    )

    dr_bin_mask = dr_edges[1:] <= dr_max
    projected_values = np.sum(values[dr_bin_mask, :], axis=0)
    projected_variances = np.sum(errors[dr_bin_mask, :] ** 2, axis=0)
    projected_errors = np.sqrt(projected_variances)

    return projected_values, projected_errors, segment_fraction_edges


def plot_segment_fraction_dr_cut_overlay(
    out,
    sample,
    channels=channels,
    dr_max=0.2,
    histname="pf_dsa_cc_cut_plane_numMatchGe1",
    sample_label=None,
    output_dir="plots/segment_fraction_dr_cut",
    output_name=None,
    figsize=(16, 12),
):
    fig, ax = plt.subplots(figsize=figsize)
    ax.text(0.0, 1.01, "Private work (CMS simulation)", transform=ax.transAxes, ha="left", va="bottom")
    reference_edges = None

    for channel in channels:
        values, errors, edges = extract_segment_fraction_projection_for_dr_cut(
            out=out,
            sample=sample,
            channel=channel,
            histname=histname,
            dr_max=dr_max,
        )

        if reference_edges is None:
            reference_edges = edges
        elif not np.allclose(reference_edges, edges):
            raise ValueError(f"Segment-fraction binning differs for {channel!r}.")

        color = channel_colors.get(channel)
        ax.stairs(
            values,
            edges,
            label=channel_labels.get(channel, channel),
            color=color,
            linestyle=channel_linestyles.get(channel, "-"),
            linewidth=2.0,
        )
        ax.errorbar(
            bin_centers(edges),
            values,
            yerr=errors,
            fmt="o",
            markersize=4,
            color=color,
            ecolor=color,
            elinewidth=1.2,
            capsize=2,
            alpha=0.85,
        )

    ax.set_xlabel("PF-DSA matched segment fraction")
    ax.set_ylabel("Yield")
    ax.grid(True, which="both", linestyle=":", alpha=0.3)
    ax.legend(frameon=False, fontsize=25)

    if sample_label is None:
        sample_label = sample

    ax.set_title(f"{sample_label}\nnumMatch >= 1, dR_outer < {dr_max}", fontsize=25)
    os.makedirs(output_dir, exist_ok=True)

    if output_name is None:
        safe_dr = str(dr_max).replace(".", "p")
        output_name = f"{sample}_segment_fraction_numMatchGe1_dRlt{safe_dr}_yield.png"

    output_path = os.path.join(output_dir, output_name)
    fig.savefig(output_path, dpi=200, bbox_inches="tight")
    plt.show()
    plt.close(fig)

    print("Saved:", output_path)
    return output_path
```
